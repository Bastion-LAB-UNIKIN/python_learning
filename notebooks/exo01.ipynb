```python
n=int(input("vous avez combien d'étudiants : "))
```

```python
with open('etudiants.txt', 'w') as c:
    for etudiant in range(n):
        nom=input(f"Quel est le nom du {etudiant+1}° etudiant :..")
        c.write(f"{nom} \n")
        
    
```

```python
with open('cotation.txt', 'w') as t:
    for etudiant in range(n):
        cote=input(f"inserez la cote du  {etudiant +1}° etudiant:..")
        t.write(f"{cote}/20 \n")
        
```
